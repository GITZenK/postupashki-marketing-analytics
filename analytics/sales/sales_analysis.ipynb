# Sales Analysis — Hackathon

Анализ существующих продаж: аудит качества данных, реконструкция заказов,
повторные покупки, продуктовые сочетания, динамика продаж и аномальные дни.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from itertools import combinations
```

```python
df = pd.read_excel("../../data/base.xlsx")

df = df.rename(columns={
    "Номер студента": "student_id",
    "Сумма": "amount",
    "Курс": "course",
    "Время": "timestamp"
})
```

## 1. Dataset Overview

```python
overview = pd.Series({
    "rows": len(df),
    "columns": df.shape[1],
    "unique_buyers": df["student_id"].nunique(),
    "unique_courses": df["course"].nunique(),
    "date_from": df["timestamp"].min(),
    "date_to": df["timestamp"].max(),
    "missing_values": df.isna().sum().sum(),
    "exact_duplicates": df.duplicated().sum()
})

overview
```

## 2. Data Quality Audit

```python
quality_summary = pd.Series({
    "missing_values": df.isna().sum().sum(),
    "exact_duplicates": df.duplicated().sum(),
    "non_positive_amounts": (df["amount"] <= 0).sum(),
    "unique_amounts": df["amount"].nunique(),
    "min_amount": df["amount"].min(),
    "max_amount": df["amount"].max(),
    "course_names_with_spaces": (
        df["course"] != df["course"].str.strip()
    ).sum()
})

quality_summary
```

## 3. Order Definition & Reconstruction

```python
order_candidates = (
    df.groupby(["student_id", "timestamp"])
    .agg(
        purchase_lines=("course", "size"),
        courses=("course", list),
        amounts=("amount", list),
        unique_amounts=("amount", "nunique")
    )
    .reset_index()
)

bundle_candidates = (
    order_candidates[order_candidates["purchase_lines"] > 1]
    .sort_values("purchase_lines", ascending=False)
)
```

```python
bundle_summary = pd.Series({
    "candidate_orders": len(bundle_candidates),
    "purchase_lines_in_candidates": bundle_candidates["purchase_lines"].sum(),
    "buyers_with_candidates": bundle_candidates["student_id"].nunique(),
    "same_amount_groups": (bundle_candidates["unique_amounts"] == 1).sum(),
    "different_amount_groups": (bundle_candidates["unique_amounts"] > 1).sum()
})

bundle_summary
```

### Assumption

Строки одного `student_id` с одинаковым `timestamp` считаем одним заказом.

В данных найдено 153 такие группы. В 152 группах значения `amount` одинаковы
между строками, а в одном случае отличаются только на 0.01 из-за округления.
Это согласуется с пакетными покупками, где несколько курсов относятся к одному
моменту оплаты.

Поэтому одна строка исходного файла не всегда равна одному заказу.


```python
df["order_id"] = (
    df.groupby(["student_id", "timestamp"])
    .ngroup()
)
```

```python
orders = (
    df.groupby("order_id")
    .agg(
        student_id=("student_id", "first"),
        timestamp=("timestamp", "first"),
        courses=("course", list),
        n_courses=("course", "size"),
        amounts=("amount", list),
        unique_amounts=("amount", "nunique")
    )
    .reset_index()
)
```

### Revenue assumption

Поле `amount` трактуем как сумму выручки, относящуюся к конкретной строке курса.
Для многокурсового заказа общая выручка считается как сумма `amount` всех его строк.

Такую трактовку поддерживает случай, где стоимость пакета распределена между
несколькими курсами с разницей в 0.01 из-за округления.


```python
orders["order_revenue"] = orders["amounts"].apply(sum)
orders["is_bundle"] = orders["n_courses"] > 1
```

```python
order_summary = pd.Series({
    "purchase_lines": len(df),
    "reconstructed_orders": len(orders),
    "unique_buyers": orders["student_id"].nunique(),
    "multi_course_orders": orders["is_bundle"].sum(),
    "multi_course_share_pct": round(orders["is_bundle"].mean() * 100, 1),
    "total_revenue": orders["order_revenue"].sum(),
    "avg_order_value": orders["order_revenue"].mean(),
    "median_order_value": orders["order_revenue"].median()
})

order_summary.apply(
    lambda x: f"{x:,.1f}" if isinstance(x, float) else x
)
```

## 4. Repeat Purchase Analysis


```python
orders = orders.sort_values(["student_id", "timestamp"])

orders["order_number"] = (
    orders.groupby("student_id")
    .cumcount() + 1
)

orders["is_repeat_order"] = orders["order_number"] > 1
```

```python
buyer_orders = (
    orders.groupby("student_id")
    .agg(
        orders_count=("order_id", "count"),
        total_revenue=("order_revenue", "sum"),
        first_order=("timestamp", "min"),
        last_order=("timestamp", "max")
    )
    .reset_index()
)

buyer_orders["is_repeat_buyer"] = buyer_orders["orders_count"] > 1
```

```python
orders["previous_order"] = (
    orders.groupby("student_id")["timestamp"]
    .shift(1)
)

orders["days_since_previous_order"] = (
    orders["timestamp"] - orders["previous_order"]
).dt.total_seconds() / 86400
```

```python
repeat_summary = pd.Series({
    "buyers": len(buyer_orders),
    "observed_repeat_buyers": buyer_orders["is_repeat_buyer"].sum(),
    "observed_repeat_buyer_share_pct": round(
        buyer_orders["is_repeat_buyer"].mean() * 100, 1
    ),
    "repeat_orders": orders["is_repeat_order"].sum(),
    "repeat_order_revenue": orders.loc[
        orders["is_repeat_order"], "order_revenue"
    ].sum(),
    "median_days_to_next_order": round(
        orders.loc[
            orders["is_repeat_order"],
            "days_since_previous_order"
        ].median(), 1
    )
})

repeat_summary
```

### Interpretation

В доступном периоде 20 из 606 покупателей совершили повторную покупку — 3.3%.

Медианное время до следующего заказа среди наблюдаемых повторных покупок составляет около 12.9 дня.

Этот показатель отражает только наблюдаемый период и не является полным repeat purchase rate:
история до 4 августа отсутствует, а покупатели в конце периода имели меньше времени для повторной покупки.

## 5. Product Combinations

```python
course_pairs = []

for courses in orders.loc[orders["is_bundle"], "courses"]:
    course_pairs.extend(combinations(sorted(courses), 2))

pair_counts = (
    pd.Series(course_pairs)
    .value_counts()
    .reset_index()
)

pair_counts.columns = ["course_pair", "orders_together"]

course_order_counts = (
    orders
    .explode("courses")
    .groupby("courses")["order_id"]
    .nunique()
)

pair_metrics = pair_counts.copy()

pair_metrics[["course_a", "course_b"]] = pd.DataFrame(
    pair_metrics["course_pair"].tolist(),
    index=pair_metrics.index
)

pair_metrics["orders_with_a"] = (
    pair_metrics["course_a"].map(course_order_counts)
)

pair_metrics["orders_with_b"] = (
    pair_metrics["course_b"].map(course_order_counts)
)

pair_metrics["lift"] = (
    pair_metrics["orders_together"] * len(orders)
    / (
        pair_metrics["orders_with_a"]
        * pair_metrics["orders_with_b"]
    )
)
```

`Lift` показывает, насколько чаще два курса встречаются в одном заказе,
чем можно было бы ожидать исходя из популярности каждого курса отдельно.
Значение выше 1 указывает на положительную связь.


```python
top_pairs = (
    pair_metrics
    .sort_values("orders_together", ascending=False)
    .head(10)
    [["course_a", "course_b", "orders_together", "lift"]]
)

top_pairs["lift"] = top_pairs["lift"].round(2)

top_pairs
```

```python
top_pairs_plot = top_pairs.copy()

top_pairs_plot["pair"] = (
    top_pairs_plot["course_a"]
    + " + "
    + top_pairs_plot["course_b"]
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_pairs_plot,
    y="pair",
    x="orders_together"
)

plt.title("Top course combinations")
plt.xlabel("Orders together")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

### Interpretation

Самые частые сочетания — `ML старт + Алгоритмы старт` (24 заказа)
и `Аналитика про + Аналитика старт` (23 заказа).

При этом высокая частота не всегда означает сильную связь.
Например, `Линейная алгебра + Мат анализ` встречается только в 5 заказах,
но имеет высокий lift, то есть эти курсы покупаются вместе значительно чаще,
чем можно было бы ожидать из их отдельной популярности.

## 6. Sales & Revenue Dynamics

```python
orders["date"] = orders["timestamp"].dt.date
```

```python
daily_sales = (
    orders.groupby("date")
    .agg(
        orders=("order_id", "count"),
        revenue=("order_revenue", "sum"),
        buyers=("student_id", "nunique")
    )
    .reset_index()
)
```

```python
plt.figure(figsize=(12, 5))

sns.lineplot(
    data=daily_sales,
    x="date",
    y="orders",
    marker="o"
)

plt.title("Daily orders")
plt.xlabel("Date")
plt.ylabel("Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))

sns.lineplot(
    data=daily_sales,
    x="date",
    y="revenue",
    marker="o"
)

plt.title("Daily revenue")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 7. Sales Spikes & Anomalies

```python
complete_days = daily_sales.iloc[:-1].copy()
```

```python
q1 = complete_days["revenue"].quantile(0.25)
q3 = complete_days["revenue"].quantile(0.75)
iqr = q3 - q1

upper_bound = q3 + 1.5 * iqr
lower_bound = max(0, q1 - 1.5 * iqr)
```

```python
spike_days = (
    complete_days[complete_days["revenue"] > upper_bound]
    [["date", "orders", "revenue"]]
    .sort_values("revenue", ascending=False)
)

spike_days
```

```python
drop_days = (
    complete_days[complete_days["revenue"] < lower_bound]
    [["date", "orders", "revenue"]]
    .sort_values("revenue")
)

drop_days
```

### Interpretation

Последний день периода исключён из анализа аномалий, так как данные за 10 сентября заканчиваются в 06:09.

Найденные всплески — это даты, которые стоит сопоставить с продающими постами, скидками, запусками и рекламными размещениями.

Сам по себе всплеск продаж не доказывает влияние рекламы.

## 9. Final Findings

### Key findings

- Исходные 795 строк соответствуют 628 реконструированным заказам и 606 уникальным покупателям.
- 153 заказа (24.4%) содержат несколько курсов, поэтому одна строка исходного файла не всегда равна одному заказу.
- Общая выручка за доступный период составляет около 5.90 млн, средний чек — около 9.4 тыс., медианный — 8.95 тыс.
- Наибольшую выручку за период принесли курсы Аналитика про (~741 тыс. ₽), ML про (~715 тыс. ₽) и AI агенты (~712 тыс. ₽).
- В наблюдаемом периоде 20 покупателей (3.3%) совершили повторную покупку. Медианное время до следующего заказа — около 12.9 дня.
- Наиболее частые сочетания курсов: ML старт + Алгоритмы старт (24 заказа) и Аналитика про + Аналитика старт (23 заказа).
- Линейная алгебра + Мат анализ имеет высокий lift (8.26), но наблюдается только в 5 заказах, поэтому вывод требует осторожности.
- По IQR обнаружены два наиболее выраженных всплеска выручки: 9 августа и 5 сентября.
- Эти даты стоит сопоставить с маркетинговой историей: продающими постами, скидками, запусками и внешними размещениями.

### Limitations

- История до 4 августа отсутствует.
- Последний день периода представлен не полностью: данные заканчиваются 10 сентября в 06:09.
- В данных нет источников пользователей, рекламных касаний и стоимости размещений.
- Поэтому по текущему датасету нельзя доказать, что всплеск продаж вызван конкретной рекламой.
- Repeat purchase rate отражает только наблюдаемый период и не является полной retention-метрикой бизнеса.
